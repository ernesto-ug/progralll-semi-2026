```python
#libreria a importar
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
#leer el archivo
datos = pd.read_csv('/content/convertor.csv', sep=';', encoding='utf-8')

print(datos["ce"])
```

```python
#Obtener los datos
c = datos['ce']
f = datos['fahrenheit']
print(c, f)
```

```python
#crear el modelo
modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(units=1)
])
```

```python
#copilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(0.1), loss='mean_squared_error')
```

```python
#entrenar el modelo
epocas = modelo.fit(c, f, epochs=200, verbose=1)
```

```python
resp = modelo.predict(np.array([[100.0]]))
print(resp)
```
