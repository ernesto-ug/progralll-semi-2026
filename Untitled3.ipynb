```python
def calcular_impuesto(monto):
    tabla = [
        (0.01,        500,       1.5, 0),
        (500.01,      1000,      1.5, 3),
        (1000.01,     2000,      3,   3),
        (2000.01,     3000,      6,   3),
        (3000.01,     6000,      9,   2),
        (6000.01,     8000,      12,  2),
        (8000.01,     18000,     15,  2),
        (18000.01,    30000,     39,  2),
        (30000.01,    60000,     63,  1),
        (60000.01,    100000,    93,  0.8),
        (100000.01,   200000,    125, 0.7),
        (200000.01,   300000,    195, 0.6),
        (300000.01,   400000,    255, 0.45),
        (400000.01,   500000,    300, 0.4),
        (500000.01,   1000000,   340, 0.30),
        (1000000.01,  99999999,  490, 0.18),
    ]
    for desde, hasta, precio, adicional in tabla:
        if desde <= monto <= hasta:
            pagar = (monto - desde) / 1000 * adicional + precio
            return round(pagar, 2)
    return None


monto = float(input("Ingrese el monto de la actividad económica: $"))
resultado = calcular_impuesto(monto)

if resultado is not None:
    print(f"El valor a pagar es: ${resultado:,.2f}")
else:
    print("El monto está fuera de los rangos establecidos.")
```

```python
"""
SISTEMA ACADÉMICO — Diccionarios/JSON + POO + funciones de usuario y predefinidas
------------------------------------------------------------------------------
Notas conceptuales:
  * JSON/Diccionarios  -> guardan datos en pares clave-valor, livianos y portables.
  * POO                -> agrupa datos (atributos) y comportamiento (métodos) en clases/objetos.
  * Funciones predefinidas (sum, len, max...) vs. funciones de usuario (definidas por nosotros).
  * Listas/diccionarios ordenan y agrupan datos; if/for/while deciden y recorren esos datos.
"""

import json
import math
from collections import Counter
from functools import reduce

# ==========================================================
# 1) DATASET — 20 estudiantes
# ==========================================================

DATOS_ESTUDIANTES = [
    {"nombre": "Samuel Alexander Portillo Ceron", "edad": 20, "tiempo_traslado_min": 80, "horas_uso_mvil": 6.0,
     "notas": {"Administración de Base de Datos II": 10.0, "Estadística Computacional": 10.0,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 10.0, "Inglés": 10.0}},
    {"nombre": "Andrea Beatriz Chavez Rivas", "edad": 20, "tiempo_traslado_min": 25, "horas_uso_mvil": 4.5,
     "notas": {"Administración de Base de Datos II": 7.0, "Estadística Computacional": 7.0,
               "Programación III": 6.5, "Estructura y Diseño de Circuitos Digitales": 7.5, "Inglés": 8.0}},
    {"nombre": "Carlos David Turcios Rivas", "edad": 20, "tiempo_traslado_min": 25, "horas_uso_mvil": 5.0,
     "notas": {"Administración de Base de Datos II": 10.0, "Estadística Computacional": 9.0,
               "Programación III": 8.5, "Estructura y Diseño de Circuitos Digitales": 9.0, "Inglés": 10.0}},
    {"nombre": "Oscar Ivan Mejia Portillo", "edad": 23, "tiempo_traslado_min": 40, "horas_uso_mvil": 7.0,
     "notas": {"Administración de Base de Datos II": 6.0, "Estadística Computacional": 6.0,
               "Programación III": 5.5, "Estructura y Diseño de Circuitos Digitales": 6.5, "Inglés": 7.0}},
    {"nombre": "Gilda Mileidy González Castro", "edad": 19, "tiempo_traslado_min": 30, "horas_uso_mvil": 5.0,
     "notas": {"Administración de Base de Datos II": 10.0, "Estadística Computacional": 7.5,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 10.0, "Inglés": 10.0}},
    {"nombre": "Sofia Nicole Aguilar Reyes", "edad": 18, "tiempo_traslado_min": 10, "horas_uso_mvil": 3.0,
     "notas": {"Administración de Base de Datos II": 9.5, "Estadística Computacional": 9.0,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 9.5, "Inglés": 10.0}},
    {"nombre": "Kimberly Michelle Castro Zelaya", "edad": 18, "tiempo_traslado_min": 20, "horas_uso_mvil": 9.0,
     "notas": {"Administración de Base de Datos II": 8.5, "Estadística Computacional": 9.0,
               "Programación III": 10.0, "Estructura y Diseño de Circuitos Digitales": 8.0, "Inglés": 9.5}},
    {"nombre": "Luis Fernando Escobar Molina", "edad": 21, "tiempo_traslado_min": 35, "horas_uso_mvil": 6.5,
     "notas": {"Administración de Base de Datos II": 7.5, "Estadística Computacional": 8.5,
               "Programación III": 8.0, "Estructura y Diseño de Circuitos Digitales": 7.0, "Inglés": 6.0}},
    {"nombre": "Yeimy Julissa Parada Chávez", "edad": 20, "tiempo_traslado_min": 45, "horas_uso_mvil": 3.0,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 9.5,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 9.0, "Inglés": 10.0}},
    {"nombre": "Karen Michelle Guardado Lopez", "edad": 19, "tiempo_traslado_min": 20, "horas_uso_mvil": 5.0,
     "notas": {"Administración de Base de Datos II": 8.5, "Estadística Computacional": 8.5,
               "Programación III": 7.5, "Estructura y Diseño de Circuitos Digitales": 8.0, "Inglés": 9.0}},
    {"nombre": "Saúl Alexander Jimenez Garcia", "edad": 21, "tiempo_traslado_min": 60, "horas_uso_mvil": 5.0,
     "notas": {"Administración de Base de Datos II": 8.5, "Estadística Computacional": 8.0,
               "Programación III": 10.0, "Estructura y Diseño de Circuitos Digitales": 8.5, "Inglés": 9.0}},
    {"nombre": "Bryan Alexander Hernandez Cruz", "edad": 24, "tiempo_traslado_min": 55, "horas_uso_mvil": 8.0,
     "notas": {"Administración de Base de Datos II": 5.0, "Estadística Computacional": 5.0,
               "Programación III": 5.5, "Estructura y Diseño de Circuitos Digitales": 6.0, "Inglés": 6.0}},
    {"nombre": "Miguel Alejandro Gómez Saravia", "edad": 19, "tiempo_traslado_min": 15, "horas_uso_mvil": 4.0,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 10.0,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 9.0, "Inglés": 9.5}},
    {"nombre": "Daniela Alejandra Ventura Sosa", "edad": 20, "tiempo_traslado_min": 18, "horas_uso_mvil": 4.0,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 9.5,
               "Programación III": 9.5, "Estructura y Diseño de Circuitos Digitales": 9.0, "Inglés": 8.5}},
    {"nombre": "Josué Stiven Funes Benavides", "edad": 19, "tiempo_traslado_min": 30, "horas_uso_mvil": 5.0,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 9.0,
               "Programación III": 9.5, "Estructura y Diseño de Circuitos Digitales": 9.0, "Inglés": 10.0}},
    {"nombre": "Marvin Ernesto Alvarado Diaz", "edad": 22, "tiempo_traslado_min": 12, "horas_uso_mvil": 2.5,
     "notas": {"Administración de Base de Datos II": 7.0, "Estadística Computacional": 7.0,
               "Programación III": 8.0, "Estructura y Diseño de Circuitos Digitales": 6.5, "Inglés": 7.5}},
    {"nombre": "Jefferson Steven Castillo Leiva", "edad": 19, "tiempo_traslado_min": 15, "horas_uso_mvil": 2.3,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 9.0,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 9.5, "Inglés": 10.0}},
    {"nombre": "Ramael Ernesto Cordova Colindres", "edad": 22, "tiempo_traslado_min": 7, "horas_uso_mvil": 13.52,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 9.0,
               "Programación III": 9.5, "Estructura y Diseño de Circuitos Digitales": 9.0, "Inglés": 10.0}},
    {"nombre": "Carlos Daniel Parada Avalos", "edad": 25, "tiempo_traslado_min": 10, "horas_uso_mvil": 5.52,
     "notas": {"Administración de Base de Datos II": 8.0, "Estadística Computacional": 8.5,
               "Programación III": 8.0, "Estructura y Diseño de Circuitos Digitales": 8.0, "Inglés": 9.0}},
    {"nombre": "Daniel Enrique Quintanilla Paniagua", "edad": 19, "tiempo_traslado_min": 15, "horas_uso_mvil": 4.0,
     "notas": {"Administración de Base de Datos II": 9.0, "Estadística Computacional": 9.5,
               "Programación III": 9.0, "Estructura y Diseño de Circuitos Digitales": 10.0, "Inglés": 10.0}},
]


# ==========================================================
# 2) CLASE Estudiante (antes iba después de Estadística)
# ==========================================================

class Estudiante:
    def __init__(self, nombre, edad, tiempo_traslado_min, horas_uso_mvil, notas):
        self.nombre = nombre
        self.edad = edad
        self.tiempo_traslado_min = tiempo_traslado_min
        self.horas_uso_mvil = horas_uso_mvil
        self.notas = notas  # dict {"Materia": nota, ...}

    def promedio_general(self):
        valores = list(self.notas.values())
        return sum(valores) / len(valores) if valores else 0.0

    def aprobo(self, nota_minima=6.0):
        return self.promedio_general() >= nota_minima

    def to_dict(self):
        return {
            "nombre": self.nombre,
            "edad": self.edad,
            "tiempo_traslado_min": self.tiempo_traslado_min,
            "horas_uso_mvil": self.horas_uso_mvil,
            "notas": self.notas,
            "promedio_general": round(self.promedio_general(), 2),
            "aprobado": self.aprobo(),
        }


# ==========================================================
# 3) CLASE Estadistica — métodos en orden distinto
# ==========================================================

class Estadistica:
    @staticmethod
    def frecuencia(datos):
        return dict(Counter(datos))

    @staticmethod
    def rango(datos):
        return (max(datos) - min(datos)) if datos else 0

    @staticmethod
    def desviacion(datos):
        return math.sqrt(Estadistica.varianza(datos))

    @staticmethod
    def varianza(datos):
        if len(datos) < 2:
            return 0.0
        m = Estadistica.media(datos)
        return sum((x - m) ** 2 for x in datos) / (len(datos) - 1)

    @staticmethod
    def moda(datos):
        if not datos:
            return []
        conteo = Counter(datos)
        top = max(conteo.values())
        return [valor for valor, veces in conteo.items() if veces == top]

    @staticmethod
    def mediana(datos):
        if not datos:
            return 0.0
        ordenados = sorted(datos)
        n = len(ordenados)
        centro = n // 2
        if n % 2 == 0:
            return (ordenados[centro - 1] + ordenados[centro]) / 2.0
        return float(ordenados[centro])

    @staticmethod
    def media(datos):
        return sum(datos) / len(datos) if datos else 0.0


# ==========================================================
# 4) CLASE SistemaAcademico
# ==========================================================

class SistemaAcademico:
    def __init__(self):
        self.estudiantes = []
        self.stats = Estadistica()

    def cargar_estudiantes(self, datos):
        self.estudiantes = list(map(
            lambda d: Estudiante(d["nombre"], d["edad"], d["tiempo_traslado_min"],
                                  d["horas_uso_mvil"], d["notas"]),
            datos
        ))

    def _linea(self, caracter="=", largo=60):
        print(caracter * largo)

    def reporte_general(self):
        edades = list(map(lambda e: e.edad, self.estudiantes))
        traslados = list(map(lambda e: e.tiempo_traslado_min, self.estudiantes))
        moviles = list(map(lambda e: e.horas_uso_mvil, self.estudiantes))

        self._linea("=")
        print("REPORTE GENERAL")
        self._linea("=")
        print(f">> Edad            | media {self.stats.media(edades):.2f} | desv. {self.stats.desviacion(edades):.2f}")
        print(f">> Traslado (min)  | media {self.stats.media(traslados):.2f} | desv. {self.stats.desviacion(traslados):.2f}")
        print(f">> Uso móvil (hrs) | media {self.stats.media(moviles):.2f} | desv. {self.stats.desviacion(moviles):.2f}")

    def reporte_campo(self, campo):
        valores = [v for v in map(lambda e: getattr(e, campo, None), self.estudiantes) if v is not None]
        if not valores:
            print("Campo no encontrado.")
            return
        self._linea("-")
        print(f"ANÁLISIS DE: {campo.upper()}")
        self._linea("-")
        print(f"Media               : {self.stats.media(valores):.2f}")
        print(f"Mediana             : {self.stats.mediana(valores):.2f}")
        print(f"Moda                : {self.stats.moda(valores)}")
        print(f"Varianza            : {self.stats.varianza(valores):.2f}")
        print(f"Desviación estándar : {self.stats.desviacion(valores):.2f}")
        print(f"Rango               : {self.stats.rango(valores):.2f}")
        print(f"Frecuencia          : {self.stats.frecuencia(valores)}")

    def resumen_por_materia(self):
        listas_materias = list(map(lambda e: list(e.notas.keys()), self.estudiantes))
        materias = sorted(set(reduce(lambda a, b: a + b, listas_materias, [])))

        self._linea("=")
        print("ANÁLISIS ESTADÍSTICO POR MATERIA")
        self._linea("=")
        for materia in materias:
            notas = [n for n in map(lambda e: e.notas.get(materia), self.estudiantes) if n is not None]
            if not notas:
                continue
            print(f"\n· {materia}")
            self._linea("-", 40)
            print(f"  Media               : {self.stats.media(notas):.2f}")
            print(f"  Mediana             : {self.stats.mediana(notas):.2f}")
            print(f"  Moda                : {self.stats.moda(notas)}")
            print(f"  Varianza            : {self.stats.varianza(notas):.2f}")
            print(f"  Desviación estándar : {self.stats.desviacion(notas):.2f}")
            print(f"  Rango               : {self.stats.rango(notas):.2f}")

    def ver_aprobados_reprobados(self, nota_minima=6.0):
        aprobados = list(filter(lambda e: e.aprobo(nota_minima), self.estudiantes))
        reprobados = list(filter(lambda e: not e.aprobo(nota_minima), self.estudiantes))

        self._linea("=")
        print("APROBADOS Y REPROBADOS")
        self._linea("=")
        print(f"\nAprobados ({len(aprobados)}):")
        for e in aprobados:
            print(f"  · {e.nombre} — {e.promedio_general():.2f}")

        print(f"\nReprobados ({len(reprobados)}):")
        if reprobados:
            for e in reprobados:
                print(f"  · {e.nombre} — {e.promedio_general():.2f}")
        else:
            print("  (ninguno)")

        total = len(self.estudiantes)
        if total:
            print(f"\nTasa de aprobación: {(len(aprobados) / total) * 100:.1f}%")

    def exportar_json(self):
        self._linea("=")
        print("ESTUDIANTES EN FORMATO JSON")
        self._linea("=")
        for e in self.estudiantes:
            print(json.dumps(e.to_dict(), ensure_ascii=False, indent=2))
            self._linea("-", 60)

    def promedios_por_estudiante(self):
        self._linea("=")
        print("PROMEDIO POR ESTUDIANTE")
        self._linea("=")
        for e in self.estudiantes:
            print(f"\n{e.nombre}")
            for materia, nota in e.notas.items():
                print(f"   {materia:<42}: {nota:.2f}")
            print(f"   {'Promedio general':<42}: {e.promedio_general():.2f}")

    def promedio_grupo(self):
        promedios = list(map(lambda e: e.promedio_general(), self.estudiantes))
        total = reduce(lambda a, b: a + b, promedios, 0)
        print(f"\nPromedio general del grupo: {(total / len(promedios)) if promedios else 0:.2f}")


# ==========================================================
# 5) MENÚ — dispatch por diccionario en vez de if/elif
# ==========================================================

def ejecutar_reporte_variable(sistema):
    print("\nVariables disponibles: 1) edad  2) tiempo_traslado_min  3) horas_uso_mvil")
    eleccion = input("Elige el número de variable: ")
    campos = {"1": "edad", "2": "tiempo_traslado_min", "3": "horas_uso_mvil"}
    if eleccion in campos:
        sistema.reporte_campo(campos[eleccion])
    else:
        print("Opción no válida.")


def menu():
    sistema = SistemaAcademico()
    sistema.cargar_estudiantes(DATOS_ESTUDIANTES)

    opciones = {
        "1": ("Reporte general (edad, traslado, uso móvil)", sistema.reporte_general),
        "2": ("Reporte de una variable específica", lambda: ejecutar_reporte_variable(sistema)),
        "3": ("Resumen estadístico por materia", sistema.resumen_por_materia),
        "4": ("Aprobados y reprobados", sistema.ver_aprobados_reprobados),
        "5": ("Exportar estudiantes a JSON", sistema.exportar_json),
        "6": ("Promedio por estudiante", sistema.promedios_por_estudiante),
        "7": ("Promedio general del grupo", sistema.promedio_grupo),
    }

    while True:
        print("\n" + "=" * 55)
        print("SISTEMA ESTADÍSTICO ACADÉMICO".center(55))
        print("=" * 55)
        for clave, (texto, _) in opciones.items():
            print(f"  {clave}. {texto}")
        print("  0. Salir")

        eleccion = input("\nElige una opción: ").strip()

        if eleccion == "0":
            print("Saliendo del sistema...")
            break
        elif eleccion in opciones:
            _, accion = opciones[eleccion]
            accion()
        else:
            print("Opción inválida, intenta de nuevo.")


if __name__ == "__main__":
    menu()
```
